## ex. 1

$$  
y = x^Tx,  \quad x \in \mathbb{R}^N 
$$

$$
\frac{dy}{dx} = 2x
$$

##### Решение

Рассмотрим $y$ как скалярное произведение вектора $x$ на самого себя. Тогда $y$ - это сумма квадратов компонент вектора $x$.

То есть $y = \sum_{i=1}^N x_i^2$

Тогда $y = \frac{\sum_{i=1}^N x_i^2}{dx_i}$, так как если брать производную по компоненте с другим индексом $i!=j$, то она равна 0.

По каждой компоненте вектора x производная квадрата компоненты $\frac{dy}{dx_i} = 2x_i$, тогда $\frac{dy}{dx} = 2x$.

## ex. 2

$$ y = tr(AB) \quad A,B \in \mathbb{R}^{N \times N} $$ 

$$
\frac{dy}{dA} = B^T
$$

##### Решение

Элементы главной диагонали для их суммы имеют следующий вид: $(AB)_{ii} = \sum_{j=1}^N A_{ij}B_{ji}$, где i - индексы строки, j - индексы столбцов (матрицы одного размера).

Тогда просуммировав все строки получим $y = \sum_{i=1}^N\sum_{j=1}^N A_{ij}B_{ji}$

Тогда при вычислении частной производной по каждому элементу матрицы $A$ на нашу функцию будет влиять только изменение элементов $A_{ij}B_{ji}$, а сами частные производные будут иметь вид $\frac{dy}{dA_{ij}} = \frac{\sum_{i=1}^N \sum_{j=1}^N A_{ij}B_{ji}}{\partial A_{ij}}$, т.е. при изменении элемента матрицы A с индексами i и j значение $y$ будет изменяться на значение элемента $B_{ji}$ (который соответствует значению элемента $B^T_{ij}$ )

Тогда мы получаем $\frac{dy}{dA}$ как сумму элементов $B_{ij}$ главной диагонали матрицы $B^T$ (так как изначально мы обозначали за i строки, за j - столбцы)

## ex. 3

$$  
y = x^TAc , \quad A\in \mathbb{R}^{N \times N}, x\in \mathbb{R}^{N}, c\in \mathbb{R}^{N} 
$$

$$
\frac{dy}{dx} = Ac
$$

$$
\frac{dy}{dA} = cx^T
$$ 

Hint for the latter (one of the ways): use *ex. 2* result and the fact 
$$
tr(ABC) = tr (CAB)
$$

##### Решение

Столбец умножаем на матрицу => получаем столбец = > умножаем на строку = > получаем скаляр

$y = x^TAc = \sum_{i=1}^N\sum_{j=1}^N x_{i}A_{ij}c_{j}$

В случае с $\frac{dy}{dx}$ получаем $\frac{dy}{dx_{i}} = \frac{\sum_{i=1}^N\sum_{j=1}^N x_{i}A_{ij}c_{j}}{\partial x_{i}} = \sum_{i=1}^N\sum_{j=1}^N A_{ij}c_{j}$, тогда $\frac{dy}{dx} = Ac$.

В случае с $\frac{dy}{dA}$ используем 3 свойства - 1) из посдказки, 2) что $tr(a^Tb)=b^Ta$, 3) $tr(A)=tr(A^T)$. Тогда $x^TAc = tr(A^Tc^Tx) = tr(Ac^Tx)$.

Взяв $c^Tx = B$, сможем использовать результат из ex.2. Тогда $\frac{dy}{dA}$ = $\frac{tr(Ac^Tx)}{dA} = (c^Tx)^T = cx^T$.

## ex. 4

Classic matrix factorization example. Given matrix $X$ you need to find $A$, $S$ to approximate $X$. This can be done by simple gradient descent iteratively alternating $A$ and $S$ updates.
$$
J = || X - AS ||_F^2  , \quad A\in \mathbb{R}^{N \times R} , \quad S\in \mathbb{R}^{R \times M}
$$
$$
\frac{dJ}{dS} = 2(A^TX - A^TAS)
$$

You may use one of the following approaches:

##### Решение

Воспользуемся первым подходом!

$J = || X - AS ||_F^2 = tr((X - AS)(X - AS)^T) =  tr(XX^T - X(AS)^T - ASX^T + AS(AS)^T) = tr(XX^T) - tr(X(AS)^T) - tr(ASX^T) + tr(AS(AS)^T) = tr(XX^T) - 2tr(ASX^T) + tr(AS(AS)^T)$.

$tr(XX^T)$ не содержит S и градиент равен 0.

$tr(ASX^T) = tr(AX^TS)$, тогда $\frac{tr(AX^TS)}{dS} = (AX^T)^T = A^TX$.

$tr(AS(AS)^T) = tr(S^TA^TAS)$, тогда тогда $\frac{tr(S^TA^TAS)}{dS} = 2A^TAS$.


Тогда $J = || X - AS ||_F^2 = -2A^TX + 2A^TAS = 2(A^TX - A^TAS)$.

$$\mu_{ij}=\frac{1}{n}\sum_{k=1}^np_{ij}^{(k)}.$$
The general standard deviation $\sigma$ and pixel-wise standard deviation $\sigma_{ij}$ is defined similarly.

Which of the following preprocessing steps will not change the performance of a Nearest Neighbor classifier that uses L1 distance? Select all that apply.

2. Subtracting the per pixel mean $\mu_{ij}$  ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu_{ij}$.)
3. Subtracting the mean $\mu$ and dividing by the standard deviation $\sigma$.
4. Subtracting the pixel-wise mean $\mu_{ij}$ and dividing by the pixel-wise standard deviation $\sigma_{ij}$.
5. Rotating the coordinate axes of the data.

*Your Answer:*
1, 2, 3, 4

*Your Explanation:*
